```python
df = df.drop(['College_ID'], axis=1)
df.head()
```

```python
df = pd.get_dummies(df, columns=['Internship_Experience', 'Placement'], drop_first=True)
```

```python
x = df.drop('Placement_Yes', axis=1)
y = df['Placement_Yes']
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.2, random_state=42
)
```

```python
# Check class imbalance before balancing
print("Before Balancing:")
print(f"Training set - Class distribution:\n{pd.Series(y_train).value_counts()}\n")
print(f"Percentage distribution:\n{pd.Series(y_train).value_counts(normalize=True) * 100}\n")
```

```python
# OPTION 1: SMOTE (Synthetic Minority Over-sampling Technique)
# Generates synthetic samples for minority class
from imblearn.over_sampling import SMOTE

smote = SMOTE(random_state=42)
x_train_smote, y_train_smote = smote.fit_resample(x_train, y_train)

print("\n=== OPTION 1: SMOTE ===")
print(f"After SMOTE balancing:\n{pd.Series(y_train_smote).value_counts()}\n")
print(f"New training set size: {len(x_train_smote)}")
print(f"Percentage distribution:\n{pd.Series(y_train_smote).value_counts(normalize=True) * 100}")
```

```python
# OPTION 2: Random Over-sampling
# Randomly duplicates minority class samples
from imblearn.over_sampling import RandomOverSampler

ros = RandomOverSampler(random_state=42)
x_train_ros, y_train_ros = ros.fit_resample(x_train, y_train)

print("\n=== OPTION 2: Random Over-sampling ===")
print(f"After Random Over-sampling:\n{pd.Series(y_train_ros).value_counts()}\n")
print(f"New training set size: {len(x_train_ros)}")
print(f"Percentage distribution:\n{pd.Series(y_train_ros).value_counts(normalize=True) * 100}")
```

```python
# OPTION 3: Random Under-sampling
# Randomly removes majority class samples
from imblearn.under_sampling import RandomUnderSampler

rus = RandomUnderSampler(random_state=42)
x_train_rus, y_train_rus = rus.fit_resample(x_train, y_train)

print("\n=== OPTION 3: Random Under-sampling ===")
print(f"After Random Under-sampling:\n{pd.Series(y_train_rus).value_counts()}\n")
print(f"New training set size: {len(x_train_rus)}")
print(f"Percentage distribution:\n{pd.Series(y_train_rus).value_counts(normalize=True) * 100}")
```

```python
# OPTION 4: SMOTE + Undersampling (Hybrid - Recommended)
# Combines synthetic generation with removal of majority class
from imblearn.pipeline import Pipeline
from imblearn.under_sampling import RandomUnderSampler
from imblearn.over_sampling import SMOTE

smote_und = Pipeline([
    ('smote', SMOTE(random_state=42)),
    ('undersampler', RandomUnderSampler(random_state=42))
])
x_train_hybrid, y_train_hybrid = smote_und.fit_resample(x_train, y_train)

print("\n=== OPTION 4: SMOTE + Under-sampling (Hybrid) ===")
print(f"After Hybrid balancing:\n{pd.Series(y_train_hybrid).value_counts()}\n")
print(f"New training set size: {len(x_train_hybrid)}")
print(f"Percentage distribution:\n{pd.Series(y_train_hybrid).value_counts(normalize=True) * 100}")
```

```python
# COMPARISON: Train models with each balancing technique and evaluate
from sklearn.metrics import precision_recall_fscore_support, roc_auc_score, confusion_matrix

print("\n" + "="*80)
print("MODEL COMPARISON WITH DIFFERENT BALANCING TECHNIQUES")
print("="*80 + "\n")

balancing_options = {
    "No Balancing": (x_train, y_train),
    "SMOTE": (x_train_smote, y_train_smote),
    "Random Over-sampling": (x_train_ros, y_train_ros),
    "Random Under-sampling": (x_train_rus, y_train_rus),
    "SMOTE + Under-sampling": (x_train_hybrid, y_train_hybrid)
}

results = {}

for technique, (X_train, y_train_bal) in balancing_options.items():
    print(f"\n{'='*40}")
    print(f"Technique: {technique}")
    print(f"{'='*40}")
    print(f"Training set size: {len(X_train)}")

    # Train Random Forest (works well with balanced data)
    model = RandomForestClassifier(random_state=42, n_estimators=100)
    model.fit(X_train, y_train_bal)

    # Predictions on test set
    y_pred = model.predict(x_test)
    y_pred_proba = model.predict_proba(x_test)[:, 1]

    # Calculate metrics
    accuracy = model.score(x_test, y_test)
    precision, recall, f1, _ = precision_recall_fscore_support(y_test, y_pred, average='weighted')
    roc_auc = roc_auc_score(y_test, y_pred_proba)
    tn, fp, fn, tp = confusion_matrix(y_test, y_pred).ravel()

    # Store results
    results[technique] = {
        'accuracy': accuracy,
        'precision': precision,
        'recall': recall,
        'f1': f1,
        'roc_auc': roc_auc,
        'tp': tp,
        'tn': tn,
        'fp': fp,
        'fn': fn
    }

    print(f"Accuracy:  {accuracy:.4f}")
    print(f"Precision: {precision:.4f}")
    print(f"Recall:    {recall:.4f}")
    print(f"F1-Score:  {f1:.4f}")
    print(f"ROC-AUC:   {roc_auc:.4f}")
    print(f"\nConfusion Matrix:")
    print(f"  True Negatives:  {tn}")
    print(f"  False Positives: {fp}")
    print(f"  False Negatives: {fn}")
    print(f"  True Positives:  {tp}")

# Summary comparison
print(f"\n\n{'='*80}")
print("SUMMARY COMPARISON")
print(f"{'='*80}\n")

comparison_df = pd.DataFrame(results).T
print(comparison_df[['accuracy', 'precision', 'recall', 'f1', 'roc_auc']].round(4))

print("\n✅ RECOMMENDATION: Choose the technique with best F1-Score (balances precision & recall)")
print("   For imbalanced data, F1-Score is often more important than Accuracy!")
```

```python
print("""
=== BALANCING TECHNIQUES GUIDE ===

1. SMOTE (Synthetic Minority Over-sampling Technique)
   - Generates synthetic samples using k-nearest neighbors
   - Best for: Small to medium datasets
   - Pros: Creates realistic synthetic samples
   - Cons: Can be slow with very large datasets

2. Random Over-sampling
   - Randomly duplicates minority class samples
   - Best for: Quick implementation
   - Pros: Simple, fast
   - Cons: Can cause overfitting due to duplicates

3. Random Under-sampling
   - Randomly removes majority class samples
   - Best for: Large datasets with lots of redundancy
   - Pros: Reduces dataset size, faster training
   - Cons: Loss of information from removed samples

4. SMOTE + Under-sampling (Hybrid) [RECOMMENDED]
   - Combines synthetic generation with selective removal
   - Best for: Most scenarios (balanced approach)
   - Pros: Preserves information while balancing
   - Cons: More parameters to tune

5. Class Weights
   - Adjusts model weights to penalize minority class errors
   - Best for: When you want to keep original data
   - Pros: No resampling, simpler approach
   - Cons: Less effective for severe imbalance

RECOMMENDATION:
Based on F1-Score comparison, choose the technique that scored highest!
F1-Score is ideal for imbalanced data as it balances precision & recall.
""")
```
